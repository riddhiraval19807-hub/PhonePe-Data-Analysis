```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/PhonePe/pulse/master/data/map/transaction/hover/country/india/2024/1.json"

data = pd.read_json(url)

print(data)
```

```python
# Extract transaction records
records = data['data'].iloc[0]

# Create an empty list
rows = []

# Extract state and transaction details
for state_info in records:
    for metric in state_info['metric']:
        rows.append({
            'State': state_info['name'],
            'Transaction_Type': metric['type'],
            'Transaction_Count': metric['count'],
            'Transaction_Amount': metric['amount']
        })

# Convert records into a DataFrame
df = pd.DataFrame(rows)

# Display first 5 rows
print(df.head())
```

```python
# Total number of transactions
total_transactions = df['Transaction_Count'].sum()

# Total transaction amount
total_amount = df['Transaction_Amount'].sum()

# Average transaction amount per transaction
average_amount = total_amount / total_transactions

# Highest transaction count in a state
highest_transactions = df['Transaction_Count'].max()

# Lowest transaction count in a state
lowest_transactions = df['Transaction_Count'].min()

# Display results
print("Total Transactions:", f"{total_transactions:,}")
print("Total Transaction Amount: ₹", f"{total_amount:,.2f}")
print("Average Transaction Amount: ₹", f"{average_amount:,.2f}")
print("Highest State Transactions:", f"{highest_transactions:,}")
print("Lowest State Transactions:", f"{lowest_transactions:,}")
```

```python
state_analysis = df.sort_values(
    by='Transaction_Count',
    ascending=False
)

print(state_analysis)
```

```python
top_5_states = state_analysis.head(5)

print("Top 5 States by Transaction Count:")
print(top_5_states[['State', 'Transaction_Count', 'Transaction_Amount']])
```

```python
df['Transaction_Percentage'] = (
    df['Transaction_Count'] / total_transactions
) * 100

print(
    df[['State', 'Transaction_Count', 'Transaction_Percentage']]
    .sort_values(by='Transaction_Percentage', ascending=False)
    .head(5)
)
```

```python
print("========== PHONEPE DATA ANALYSIS ==========")

print("Total States:", df['State'].nunique())

print("Total Transactions:", f"{total_transactions:,}")

print("Total Transaction Amount: ₹", f"{total_amount:,.2f}")

print("Average Transaction Amount: ₹", f"{average_amount:,.2f}")

print("Top Performing State:", state_analysis.iloc[0]['State'])

print("Lowest Performing State:", state_analysis.iloc[-1]['State'])

print("===========================================")
```

```python
# Convert transaction amount column into NumPy array
amounts = np.array(df['Transaction_Amount'])

# Calculate statistics using NumPy
print("Average Amount:", np.mean(amounts))
print("Maximum Amount:", np.max(amounts))
print("Minimum Amount:", np.min(amounts))
print("Standard Deviation:", np.std(amounts))
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nNegative Transaction Counts:")
print((df['Transaction_Count'] < 0).sum())

print("\nNegative Transaction Amounts:")
print((df['Transaction_Amount'] < 0).sum())
```
